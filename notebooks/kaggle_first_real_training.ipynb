# Myownai — أول تدريب حقيقي على نصوص عربية وإنجليزية

هذه الخطوة تنزّل عينة محدودة من مقالات ويكيبيديا العربية والإنجليزية، تدرب Tokenizer خاصًا بالمشروع من الصفر، ثم تدرب Transformer بأوزان عشوائية على Tesla T4 أو أي GPU متاح.

**قبل التشغيل:** تأكد من Settings → Accelerator → GPU، ومن تفعيل Internet. شغّل الخلية الوحيدة بالأسفل.

**تنبيه الترخيص:** بطاقة مجموعة Wikimedia/Wikipedia تعلن CC BY-SA 3.0 وGFDL. يحتفظ السكريبت ببيانات المصدر والإسناد لكل مقالة. لا تنشر ملفات corpus أو checkpoint قبل مراجعة التزامات ShareAlike/GFDL. هذه تجربة أولية صغيرة وليست مساعدًا بمستوى Claude.

```python
import os
import shutil
import subprocess
import sys
from pathlib import Path

# Start outside the repo so a previous folder can be safely replaced.
os.chdir('/kaggle/working')
repo = Path('/kaggle/working/Myownai')
if repo.exists():
    shutil.rmtree(repo)
subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/Eslamrady42/Myownai.git', str(repo)], check=True)
os.chdir('/kaggle/working')
print('Repository downloaded:', repo)

# This downloads a limited balanced corpus, trains a tokenizer and a small model from random weights,
# runs Arabic/English generation, and saves a ZIP in /kaggle/working.
subprocess.run([
    sys.executable,
    str(repo / 'scripts' / 'train_wikipedia_micro.py'),
    '--require-gpu',
    '--articles-per-language', '1500',
    '--steps', '300',
    '--batch-size', '8',
    '--vocab-size', '4096',
], cwd='/kaggle/working', check=True)
```
